## Executive Summary
This study evaluates the patterns of life expectancy across the globe using the World Bank's Health, Nutrition and Population Statistics dataset. Specifically, the study evaluates the changes in life expectancy from 2014 to 2020 and any variations depending on geographical region, gender, and income group. For the data cleaning, analysing and visualisation process, I have employed the Python programming language especially Pandas and Matplotlib libraries. It has been noted that the average life expectancy increased from 71.75 years in 2014 to 72.98 years in 2020 in 199 countries and economies in which data is available for both years. Of the 199 economies, 169 have reported an increase while 30 have reported a decrease in life expectancy in 2020 relative to 2014. Regionally, North America emerged as the geographical region with the highest average life expectancy, being 80.36 years compared to the average life expectancy of 63.28 years in Sub-Saharan Africa. There has been noted to be about 4.89 years difference between the female and male life expectancy.

## Introduction
The life expectancy of people is one of the most significant health indicators of the population. This term is used to indicate the average age that people are going to live till and serves as a means of evaluation for healthcare facilities, economic situation, nutrition, etc. Variations in the life expectancy among nations may demonstrate differences in availability of medicine, economic situation, and nutrition.

In this research, the World Bank dataset on Health, Nutrition and Population statistics is used to determine the life expectancy rates in different economies of the world. The first purpose of this study is to conduct an analysis of the trends in life expectancy rate from 2014 to 2020. The difference in life expectancy rates due to the geography of the area will be determined.

This graph illustrates the five nations/economies which witnessed the highest improvements in their life expectancies from 2014 to 2020. The years added by each nation during the given time period have been compared to show where the maximum improvement was observed.

Is average life expectancy higher for females or males across countries and economies in 2020?

The current analysis aims to find the disparities in life expectancy between women and men in various countries and economies in 2020. There are separate World Bank indicators for the life expectancy of women and men at birth in the dataset used in this analysis; therefore, they can be analyzed separately. The required information is chosen from the dataset, and the average life expectancy is calculated for each of the genders based on the provided information. Then, the average life expectancies are compared to see the disparity between women's and men's life expectancy as well as to understand who has a higher average life expectancy.

```python
# Find the female and male life expectancy indicators
# The search is case-insensitive and ignores missing indicator names
gender_indicators = data[
    data['Indicator Name'].str.contains(
        'Life expectancy at birth',
        case=False,
        na=False
    )
]

# Display the available indicator names
print(gender_indicators['Indicator Name'].unique())
```

```python
# Select female life expectancy records
female_data = data[
    data['Indicator Name'] == 'Life expectancy at birth, female (years)'
][['Country Code', '2020']].copy()

# Rename the 2020 column to identify female values
female_data = female_data.rename(
    columns={'2020': 'Female'}
)

# Select male life expectancy records
male_data = data[
    data['Indicator Name'] == 'Life expectancy at birth, male (years)'
][['Country Code', '2020']].copy()

# Rename the 2020 column to identify male values
male_data = male_data.rename(
    columns={'2020': 'Male'}
)

# Match female and male records using their country codes
gender_data = female_data.merge(
    male_data,
    on='Country Code',
    how='inner'
)

# Remove records with missing female or male values
gender_data = gender_data.dropna(
    subset=['Female', 'Male']
)

# Display the first five records
print(gender_data.head())
```

```python
# Keep only country/economy codes listed in the World Bank metadata
gender_countries = gender_data[
    gender_data['Country Code'].isin(country_codes)
].copy()

# Calculate the average female life expectancy
average_female = gender_countries['Female'].mean()

# Calculate the average male life expectancy
average_male = gender_countries['Male'].mean()

# Calculate the difference between female and male averages
average_difference = average_female - average_male

# Display the results rounded to two decimal places
print("Average female life expectancy:", round(average_female, 2))
print("Average male life expectancy:", round(average_male, 2))
print("Female advantage in years:", round(average_difference, 2))

# Display the number of countries/economies included
print("Number of countries/economies:", len(gender_countries))
```

From the results, it is clear that females have a higher life expectancy average than males in the year 2020. Females have an average life expectancy of 75.47 years whereas the average life expectancy of males is 70.57 years. Therefore, there is a difference of about 4.89 years between the two genders based on life expectancy statistics in the 199 countries and economies under consideration.

How does average life expectancy differ across World Bank income groups in 2020?

The following analysis evaluates the variance of average life expectancy across the four income levels in the World Bank classification in 2020. Country and economy classifications include the high, upper middle, lower middle and low income countries according to the World Bank country metadata. The average life expectancies across the income levels are evaluated to determine whether there is an association between economic classification and life expectancy.

This comparison matters because income levels can influence access to healthcare, medical facilities, nutritious food, sanitation and other essential services. Examining life expectancy across income groups helps identify potential inequalities in health outcomes and highlights whether people living in higher-income economies generally experience longer life expectancy than those in lower-income economies.

```python
# Display all column names in the country metadata
# This helps us identify the income-group information
print(countries.columns.tolist())
```

```python
# Combine life expectancy records with country income-group information
income_data = life_countries.merge(
    countries[['Country Code', 'Income Group']],
    on='Country Code',
    how='left'
)

# Remove records with missing 2020 values or income groups
income_2020 = income_data.dropna(
    subset=['2020', 'Income Group']
).copy()

# Group the data by income group and calculate average life expectancy
income_averages = income_2020.groupby(
    'Income Group'
)['2020'].mean()

# Sort the results from highest to lowest
income_averages = income_averages.sort_values(
    ascending=False
)

# Display the average life expectancy for each income group
print(income_averages.round(2))
```

```python
# Import matplotlib for creating the chart
import matplotlib.pyplot as plt

# Create a horizontal bar chart of income-group averages
plt.barh(
    income_averages.index,
    income_averages.values
)

# Place the highest income group at the top
plt.gca().invert_yaxis()

# Display the average value beside each bar
for index, value in enumerate(income_averages.values):
    plt.text(
        value + 0.3,
        index,
        f'{value:.2f}',
        va='center'
    )

# Add the chart title and axis labels
plt.title('Average Life Expectancy by Income Group (2020)')
plt.xlabel('Average Life Expectancy (Years)')
plt.ylabel('Income Group')

# Adjust spacing and display the chart
plt.tight_layout()
plt.show()
```

The statistics clearly indicate the disparities in the mean life expectancy levels among the four categories of World Bank economies. The high income category registered the highest mean life expectancy of 79.82 years, while the low income category had the lowest mean life expectancy level of 62.95 years. This was a disparity of 16.87 years. On the other hand, the upper middle income economies had a mean life expectancy level of 73.55 years and the lower middle economies had a mean of 69.20 years.
